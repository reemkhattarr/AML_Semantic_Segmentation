```python
from detectron2.data import DatasetCatalog, MetadataCatalog
import os, glob, cv2

LOVEDA_CLASSES = ["background","building","road","water","barren","forest","agriculture"]

def get_loveda_semseg_dicts(root, split):
    dataset_dicts = []
    for area in ["Urban", "Rural"]:
        img_dir = os.path.join(root, split, area, "images_png")
        mask_dir = os.path.join(root, split, area, "masks_png")
        img_paths = sorted(glob.glob(os.path.join(img_dir, "*.png")))
        for img_path in img_paths:
            file = os.path.basename(img_path)
            sem = os.path.join(mask_dir, file)
            im = cv2.imread(img_path)
            h, w = im.shape[:2]
            dataset_dicts.append({
                "file_name": img_path,
                "sem_seg_file_name": sem,
                "height": h,
                "width": w,
            })
    return dataset_dicts

def register_loveda(root="/content/drive/MyDrive/AML_Semantic_Segmentation/data/LoveDA"):
    for split in ["Train", "Val"]:
        name = f"loveda_{split.lower()}"
        DatasetCatalog.register(name, lambda split=split: get_loveda_semseg_dicts(root, split))
        MetadataCatalog.get(name).set(
            stuff_classes=LOVEDA_CLASSES,
            ignore_label=0,
            evaluator_type="sem_seg",
        )

register_loveda()
print("Registrati:", DatasetCatalog.list())
```

```python
from detectron2.data import DatasetCatalog, MetadataCatalog

for name in ["loveda_train", "loveda_val"]:
    if name in DatasetCatalog.list():
        # rimuovi dal catalogo dataset
        DatasetCatalog.remove(name)
    # rimuovi i metadati (se supportato)
    if hasattr(MetadataCatalog, "remove"):
        try:
            MetadataCatalog.remove(name)
        except Exception:
            pass
    else:
        # fallback: azzera i metadati (se remove non esiste)
        try:
            MetadataCatalog.get(name).set(thing_classes=[], stuff_classes=[])
        except Exception:
            pass
```
